```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_duplicate = df.duplicated(subset=['Ort','Straße','Fachrichtung','Name'],keep=False)
```

```python
df_m = df[df_duplicate]
```

```python
df_ohen_duplicate = df[~df_duplicate]
```

```python
df_ohen_duplicate
```

```python
df_nun = df_ohen_duplicate[df_ohen_duplicate[['Name', 'Ort']].notnull().all(axis=1)]
```

```python
df_nun
```

```python
df_nun
```

```python
df_anschauen = df_nun.groupby(['Ort','Fachrichtung']).agg(
    anzahl = ('ID-Nr','count'),
    gesamt_wert = ('Gesamt\nkumuliert\n2025\nNovember','sum'),
    gesamt_stuck = ('Gesamt\nkumuliert in Stück\n2025\nNovember','sum')
).reset_index()
```

```python
df_anschauen
```

```python
df_top10 = df_anschauen.sort_values(by='gesamt_wert', ascending=False).head(20)
```

```python
df_top10['Label'] = df_top10['Ort'] + " - " + df_top10['Fachrichtung']
```

```python
sns.barplot(data=df_top10, x='gesamt_wert', y='Label', palette='viridis')

plt.title('Top 10 Standorte/Fachrichtungen nach Gesamtwert (November 2025)', fontsize=15)
plt.xlabel('Kumulierter Gesamtwert (€)', fontsize=12)
plt.ylabel('Ort & Fachrichtung', fontsize=12)

# Das Diagramm anzeigen
plt.tight_layout()
plt.show()
```
